````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
#1
cities_clean = [city.strip().capitalize() for city in cities]
print(cities_clean)

#2
amounts_int = [int(num) for num in amounts]
print(amounts_int)

#3
cnt_checks = len(amounts_int)
amt_checks = sum(amounts_int)
avg_checks = amt_checks/cnt_checks
print(avg_checks)

#4
unpaid_checks = []
for i in range(cnt_checks):
    if paid[i]:
        continue
    else:
        unpaid_checks.append(amounts_int[i])

amt_unpaid_checks = sum(unpaid_checks)
print(unpaid_checks)

#5
zipped = list(zip(cities_clean, amounts_int, paid))

print(f"{'город'} | {'сумма'} | {'статус'}")
for row in zipped:
    print(f"{row[0]} | {row[1]} | {row[2]}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт
#6
amt_paid_almaty_astana = 0
for i in range(cnt_checks):
    if paid[i] and cities_clean[i] in ('Almaty', 'Astana'):
        amt_paid_almaty_astana += amounts_int[i]
print(amt_paid_almaty_astana)

#7
index_of_biggest = amounts_int.index(max(amounts_int))
print(index_of_biggest)

#8
cnt_weekend = len([day for day in days if day in ('сб', 'вс')])
print(cnt_weekend/len(days)*100)

#9
amt_student = 0
start_step = 15000
target_amt = 400000
cnt_month = 0
while amt_student < target_amt:
    amt_student += start_step
    start_step += 1000
    cnt_month += 1

print(cnt_month)
````

## [Челлендж]
10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.
11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.
12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.
```
пн | ████  20490
вт | ...
```
13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
# Челлендж
#10
unique_cities = []
for city in cities_clean:
    if city not in unique_cities:
        unique_cities.append(city)
        
revenue_by_city = [0] * len(unique_cities)

for i in range(len(unique_cities)):
    for k in range(len(cities_clean)):
        if cities_clean[k] == unique_cities[i]:
            revenue_by_city[i] += amounts_int[k]

zip_unique = list(zip(unique_cities,revenue_by_city))

sorted_zip_unique = sorted(zip_unique, key=lambda x: x[1], reverse=True)
unique_cities, revenue_by_city = zip(*sorted_zip_unique)
for i in range(len(unique_cities)):
    print(unique_cities[i], revenue_by_city[i])

#11
sorted_amounts_int = sorted(amounts_int)
middle = len(sorted_amounts_int) // 2
median = (sorted_amounts_int[middle] + sorted_amounts_int[middle]) / 2
print(median)

#12
amt_by_days = {'пн': 0, 'вт': 0, 'ср': 0, 'чт': 0, 'пт': 0, 'сб': 0, 'вс': 0}
for i in range(len(days)):
    amt_by_days[days[i]] = amt_by_days.get(days[i], 0) + amounts_int[i]

for day, amt_day in amt_by_days.items():
    cnt_blocks = amt_day // 5000
    print(f"{day} | {cnt_blocks * '█'} {amt_day}")
````
